# F-16 ground vibration test

See https://data.4tu.nl/articles/_/12954911 for more information on the dataset.

This example fits only a linear model. It demonstrates the `enforce_stability` option in BLA optimization and how to initialize a `freq-statespace` data object from a BLA estimated with the `best-linear-approximation` package.

```python
import best_linear_approximation as bla
import freq_statespace as fss

raw_data = bla.dataloader.load_f16()["F16Data_FullMSine_Level7.mat"]

bla_estimate = bla.robust.noisy_input(raw_data.u, raw_data.y, raw_data.fs)

data = fss.create_data_object_from_bla(bla_estimate)
```

```python
# Estimate a BLA without enforcing stability
nx = 16  # Number of states
bla_subspace = fss.lin.subspace_id(data, nx)

bla = fss.lin.optimize(bla_subspace, data, print_every=25)  # low cost yet unstable model

# Optimize the same initial model with stability enforced
bla = fss.lin.optimize(bla_subspace, data, enforce_stability=True, print_every=25)  # low cost and stable model
```
