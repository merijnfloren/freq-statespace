```python
import best_linear_approximation as bla
import numpy as np
import freq_statespace as fss

raw_data = bla.dataloader.load_f16()["F16Data_FullMSine_Level7.mat"]

r = np.mean(raw_data.r, axis=-1)

bla_estimate = bla.robust.known_reference(r, raw_data.u, raw_data.y, raw_data.fs)

data = fss.create_data_object_from_bla(bla_estimate)
```

```python
# Step 1: BLA estimation
nx = 16  # state dimension
bla = fss.lin.subspace_id(data, nx)
bla = fss.lin.optimize(bla, data, enforce_stability=True)
```
